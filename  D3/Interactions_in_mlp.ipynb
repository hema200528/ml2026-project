```python
# Load validation splits (for tuning without touching test)
pneumonia_val = PneumoniaMNIST(split="val", download=True)
blood_val     = BloodMNIST(split="val", download=True)

# Pneumonia arrays
Xp_val = pneumonia_val.imgs.reshape(len(pneumonia_val.imgs), -1) / 255.0
yp_val = pneumonia_val.labels.flatten()

# Blood arrays
Xb_val = blood_val.imgs.reshape(len(blood_val.imgs), -1) / 255.0
yb_val = blood_val.labels.flatten()

# Standardise: subtract mean, divide by std (computed from TRAIN only)
def standardise(X_tr, *others):
    mean, std = X_tr.mean(0), X_tr.std(0) + 1e-8
    return [(X - mean) / std for X in (X_tr, *others)]

Xp_tr_s, Xp_val_s, Xp_te_s = standardise(Xp_tr, Xp_val, Xp_te)
Xb_tr_s, Xb_val_s, Xb_te_s = standardise(Xb_tr, Xb_val, Xb_te)

print("Standardised. Pneumonia train:", Xp_tr_s.shape, "| Blood train:", Xb_tr_s.shape)
```

```python
import copy

def train_good_mlp(X_tr, y_tr, X_val, y_val, input_size, num_classes,
                   h1=512, h2=256, dropout=0.3, lr=1e-3, batch_size=128,
                   max_epochs=100, patience=10):
    torch.manual_seed(42)
    loader = DataLoader(
        TensorDataset(torch.FloatTensor(X_tr), torch.LongTensor(y_tr)),
        batch_size=batch_size, shuffle=True)

    model = nn.Sequential(
        nn.Linear(input_size, h1), nn.ReLU(), nn.Dropout(dropout),
        nn.Linear(h1, h2), nn.ReLU(), nn.Dropout(dropout),
        nn.Linear(h2, num_classes))
    opt = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
    loss_fn = nn.CrossEntropyLoss()

    Xv = torch.FloatTensor(X_val)
    best_acc, best_state, wait, history = 0, None, 0, []

    for epoch in range(max_epochs):
        model.train()
        for xb, yb in loader:
            opt.zero_grad()
            loss_fn(model(xb), yb).backward()
            opt.step()

        model.eval()
        with torch.no_grad():
            val_preds = model(Xv).argmax(1).numpy()
        val_acc = accuracy_score(y_val, val_preds)
        history.append(val_acc)

        if val_acc > best_acc:
            best_acc, best_state, wait = val_acc, copy.deepcopy(model.state_dict()), 0
        else:
            wait += 1
            if wait >= patience:
                print(f"  early stop at epoch {epoch+1}")
                break

        if (epoch+1) % 5 == 0:
            print(f"  epoch {epoch+1}: val acc {val_acc:.4f} (best {best_acc:.4f})")

    model.load_state_dict(best_state)
    return model, history, best_acc
```

```python
print("Training MLP on Pneumonia...")
mlp_p, hist_p, best_p = train_good_mlp(Xp_tr_s, yp_tr, Xp_val_s, yp_val,
                                       input_size=784, num_classes=2)

# Final test-set evaluation
mlp_p.eval()
with torch.no_grad():
    preds = mlp_p(torch.FloatTensor(Xp_te_s)).argmax(1).numpy()
mlp_p_acc = accuracy_score(yp_te, preds)
mlp_p_f1 = f1_score(yp_te, preds, average='macro')
print(f"\nPneumonia MLP — Test Accuracy {mlp_p_acc:.4f} | Macro F1 {mlp_p_f1:.4f}")
```

```python
!pip install medmnist -q
import numpy as np, torch, copy, time, warnings
import torch.nn as nn, torch.optim as optim
import torch.nn.utils.prune as prune
from torch.utils.data import DataLoader, TensorDataset
from medmnist import PneumoniaMNIST, BloodMNIST
from sklearn.metrics import accuracy_score, f1_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
warnings.filterwarnings("ignore")

SEED=42; np.random.seed(SEED); torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device", device)

def load(ds):
    tr,va,te=(ds(split=s,download=True) for s in ["train","val","test"])
    def prep(d):
        X=d.imgs.reshape(len(d.imgs),-1).astype(np.float32)/255.0
        return X, d.labels.flatten().astype(np.int64)
    return prep(tr),prep(va),prep(te)

def standardise(Xtr,*rest):
    m,s=Xtr.mean(0),Xtr.std(0)+1e-8
    return [(X-m)/s for X in (Xtr,*rest)]

(Xp_tr,yp_tr),(Xp_va,yp_va),(Xp_te,yp_te)=load(PneumoniaMNIST)
(Xb_tr,yb_tr),(Xb_va,yb_va),(Xb_te,yb_te)=load(BloodMNIST)
Xp_tr,Xp_va,Xp_te=standardise(Xp_tr,Xp_va,Xp_te)
Xb_tr,Xb_va,Xb_te=standardise(Xb_tr,Xb_va,Xb_te)
print("loaded", Xp_tr.shape, Xb_tr.shape)
```

```python
class MLP(nn.Module):
    def __init__(self,in_dim,n_classes,h1=512,h2=256,p=0.3):
        super().__init__()
        self.fc1=nn.Linear(in_dim,h1); self.fc2=nn.Linear(h1,h2); self.fc3=nn.Linear(h2,n_classes)
        self.drop=nn.Dropout(p)
    def forward(self,x):
        x=self.drop(torch.relu(self.fc1(x)))
        x=self.drop(torch.relu(self.fc2(x)))
        return self.fc3(x)

def train_mlp(Xtr,ytr,Xva,yva,in_dim,n_classes,batch_size=128,lr=1e-3,dropout=0.3,max_epochs=80,patience=10):
    torch.manual_seed(SEED)
    model=MLP(in_dim,n_classes,p=dropout).to(device)
    opt=optim.Adam(model.parameters(),lr=lr,weight_decay=1e-4); lossf=nn.CrossEntropyLoss()
    loader=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=batch_size,shuffle=True)
    Xva_t=torch.tensor(Xva).to(device)
    best,best_state,wait=0,None,0
    for ep in range(max_epochs):
        model.train()
        for xb,yb in loader:
            xb,yb=xb.to(device),yb.to(device)
            opt.zero_grad(); lossf(model(xb),yb).backward(); opt.step()
        model.eval()
        with torch.no_grad():
            va=accuracy_score(yva,model(Xva_t).argmax(1).cpu().numpy())
        if va>best: best,best_state,wait=va,copy.deepcopy(model.state_dict()),0
        else:
            wait+=1
            if wait>=patience: break
    model.load_state_dict(best_state)
    return model

def evaluate(model,X,y):
    model.eval()
    with torch.no_grad():
        p=model(torch.tensor(X).to(device)).argmax(1).cpu().numpy()
    return accuracy_score(y,p), f1_score(y,p,average='macro')
```

```python
def classical(Xtr,ytr,Xte,yte,name):
    out={}
    for label,clf in [("LogReg",LogisticRegression(max_iter=1000,random_state=SEED)),
                      ("RandomForest",RandomForestClassifier(n_estimators=100,random_state=SEED,n_jobs=-1)),
                      ("RBF_SVM",SVC(kernel='rbf',random_state=SEED))]:
        clf.fit(Xtr,ytr); pr=clf.predict(Xte)
        out[label]=(accuracy_score(yte,pr),f1_score(yte,pr,average='macro'))
        print(f"  {name} {label}: acc {out[label][0]:.4f} f1 {out[label][1]:.4f}")
    return out

print("Pneumonia classical:"); cls_p=classical(Xp_tr,yp_tr,Xp_te,yp_te,"P")
print("Blood classical:");     cls_b=classical(Xb_tr,yb_tr,Xb_te,yb_te,"B")
```

```python
mlp_p=train_mlp(Xp_tr,yp_tr,Xp_va,yp_va,784,2,dropout=0.5,lr=3e-4)
mlp_b=train_mlp(Xb_tr,yb_tr,Xb_va,yb_va,2352,8,dropout=0.3,lr=1e-3)
print("MLP Pneumonia", evaluate(mlp_p,Xp_te,yp_te))
print("MLP Blood", evaluate(mlp_b,Xb_te,yb_te))
```

```python
def prune_model(model,amount):
    pm=copy.deepcopy(model)
    for _,m in pm.named_modules():
        if isinstance(m,nn.Linear):
            prune.l1_unstructured(m,'weight',amount=amount); prune.remove(m,'weight')
    return pm

def sparsity(model):
    tot=z=0
    for m in model.modules():
        if isinstance(m,nn.Linear): tot+=m.weight.numel(); z+=(m.weight==0).sum().item()
    return z/tot

def finetune(model,Xtr,ytr,epochs=10,lr=5e-4,bs=128):
    model.train(); opt=optim.Adam(model.parameters(),lr=lr,weight_decay=1e-4); lossf=nn.CrossEntropyLoss()
    loader=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=bs,shuffle=True)
    for _ in range(epochs):
        for xb,yb in loader:
            xb,yb=xb.to(device),yb.to(device)
            opt.zero_grad(); lossf(model(xb),yb).backward(); opt.step()
    return model

def prune_study(model,Xtr,ytr,Xte,yte,name,levels=(0.1,0.3,0.5,0.7,0.8,0.9)):
    print(f"\n{name} pruning:")
    print(f"{'sparsity':>9}{'pre_acc':>9}{'post_acc':>9}")
    rows=[]
    for lvl in levels:
        pm=prune_model(model,lvl); pre=evaluate(pm,Xte,yte)[0]
        pm=finetune(pm,Xtr,ytr); post=evaluate(pm,Xte,yte)[0]
        rows.append((lvl,pre,post)); print(f"{lvl:>9.0%}{pre:>9.3f}{post:>9.3f}")
    return rows

pr_p=prune_study(mlp_p,Xp_tr,yp_tr,Xp_te,yp_te,"Pneumonia")
pr_b=prune_study(mlp_b,Xb_tr,yb_tr,Xb_te,yb_te,"Blood")
```

```python
def quantize(model,bits):
    qm=copy.deepcopy(model)
    with torch.no_grad():
        for m in qm.modules():
            if isinstance(m,nn.Linear):
                w=m.weight.data; lo,hi=w.min(),w.max()
                if hi>lo:
                    lv=2**bits-1; sc=(hi-lo)/lv
                    m.weight.copy_(torch.round((w-lo)/sc)*sc+lo)
    return qm

def quant_study(model,Xte,yte,name,bits=(8,6,4,3,2)):
    print(f"\n{name} quantization:")
    rows=[(32,)+evaluate(model,Xte,yte)]
    for b in bits: rows.append((b,)+evaluate(quantize(model,b),Xte,yte))
    for b,a,f in rows: print(f"  {b:>2}-bit acc {a:.3f} f1 {f:.3f}")
    return rows

q_p=quant_study(mlp_p,Xp_te,yp_te,"Pneumonia")
q_b=quant_study(mlp_b,Xb_te,yb_te,"Blood")
```

```python
def coreset_study(Xtr,ytr,Xva,yva,Xte,yte,in_dim,n_classes,name,dropout,lr,fracs=(0.25,0.5,1.0)):
    print(f"\n{name} coreset (random subset):")
    rows=[]
    for f in fracs:
        n=int(len(Xtr)*f); idx=np.random.RandomState(SEED).choice(len(Xtr),n,replace=False)
        m=train_mlp(Xtr[idx],ytr[idx],Xva,yva,in_dim,n_classes,dropout=dropout,lr=lr)
        acc=evaluate(m,Xte,yte)[0]; rows.append((f,acc)); print(f"  {f:>4.0%} data: acc {acc:.3f}")
    return rows

co_b=coreset_study(Xb_tr,yb_tr,Xb_va,yb_va,Xb_te,yb_te,2352,8,"Blood",0.3,1e-3)
```

```python
# ============================================================
# D3-1: ORDER INTERACTION (prune→quant vs quant→prune)
# ============================================================
def prune_then_quant(model,Xtr,ytr,amount,bits,finetune_first=True):
    pm=prune_model(model,amount)
    if finetune_first: pm=finetune(pm,Xtr,ytr)
    pm=quantize(pm,bits)
    return pm

def quant_then_prune(model,Xtr,ytr,amount,bits):
    qm=quantize(model,bits)
    pm=prune_model(qm,amount)
    pm=finetune(pm,Xtr,ytr)   # heal after pruning
    return pm

def order_study(model,Xtr,ytr,Xte,yte,name,amount=0.8,bits=4):
    pq=prune_then_quant(model,Xtr,ytr,amount,bits); a1=evaluate(pq,Xte,yte)
    qp=quant_then_prune(model,Xtr,ytr,amount,bits); a2=evaluate(qp,Xte,yte)
    print(f"\n{name} (prune {amount:.0%} + {bits}-bit):")
    print(f"  prune→quant : acc {a1[0]:.3f} f1 {a1[1]:.3f}")
    print(f"  quant→prune : acc {a2[0]:.3f} f1 {a2[1]:.3f}")
    return a1,a2

order_p=order_study(mlp_p,Xp_tr,yp_tr,Xp_te,yp_te,"Pneumonia")
order_b=order_study(mlp_b,Xb_tr,yb_tr,Xb_te,yb_te,"Blood")
```

```python
# ============================================================
# D3: FAIR ORDER COMPARISON — fine-tune after every step
# ============================================================
def finetune_keep_quant(model,Xtr,ytr,bits=None,epochs=10,lr=5e-4,bs=128):
    """Fine-tune; if bits given, re-quantize after each epoch so weights stay quantized."""
    model.train(); opt=optim.Adam(model.parameters(),lr=lr,weight_decay=1e-4); lossf=nn.CrossEntropyLoss()
    loader=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=bs,shuffle=True)
    for _ in range(epochs):
        for xb,yb in loader:
            xb,yb=xb.to(device),yb.to(device)
            opt.zero_grad(); lossf(model(xb),yb).backward(); opt.step()
        if bits is not None:
            with torch.no_grad():
                for m in model.modules():
                    if isinstance(m,nn.Linear):
                        w=m.weight.data; lo,hi=w.min(),w.max()
                        if hi>lo:
                            lv=2**bits-1; sc=(hi-lo)/lv
                            m.weight.copy_(torch.round((w-lo)/sc)*sc+lo)
    return model

def fair_prune_then_quant(model,Xtr,ytr,amount,bits):
    pm=prune_model(model,amount)              # prune
    pm=finetune(pm,Xtr,ytr)                    # heal 1
    pm=quantize(pm,bits)                       # quantize
    pm=finetune_keep_quant(pm,Xtr,ytr,bits)    # heal 2 (stays quantized)
    return pm

def fair_quant_then_prune(model,Xtr,ytr,amount,bits):
    qm=quantize(model,bits)                    # quantize
    qm=finetune_keep_quant(qm,Xtr,ytr,bits)    # heal 1 (stays quantized)
    pm=prune_model(qm,amount)                  # prune
    pm=finetune_keep_quant(pm,Xtr,ytr,bits)    # heal 2 (stays quantized)
    return pm

def fair_order_study(model,Xtr,ytr,Xte,yte,name,combos):
    print(f"\n{name} — FAIR order (fine-tune after every step)")
    print(f"{'prune':>7}{'bits':>6}{'P→Q':>9}{'Q→P':>9}{'diff':>8}")
    rows=[]
    for amount,bits in combos:
        a1=evaluate(fair_prune_then_quant(model,Xtr,ytr,amount,bits),Xte,yte)[0]
        a2=evaluate(fair_quant_then_prune(model,Xtr,ytr,amount,bits),Xte,yte)[0]
        rows.append((amount,bits,a1,a2,a1-a2))
        print(f"{amount:>6.0%}{bits:>6}{a1:>9.3f}{a2:>9.3f}{a1-a2:>+8.3f}")
    return rows

combos=[(0.70,3),(0.90,3),(0.70,2),(0.90,2)]
fair_p=fair_order_study(mlp_p,Xp_tr,yp_tr,Xp_te,yp_te,"Pneumonia",combos)
fair_b=fair_order_study(mlp_b,Xb_tr,yb_tr,Xb_te,yb_te,"Blood",combos)
```

```python
!pip install medmnist -q
import numpy as np, torch, copy, warnings
import torch.nn as nn, torch.optim as optim
import torch.nn.utils.prune as prune
from torch.utils.data import DataLoader, TensorDataset
from medmnist import PneumoniaMNIST, BloodMNIST
from sklearn.metrics import accuracy_score, f1_score
warnings.filterwarnings("ignore")

SEED=42; np.random.seed(SEED); torch.manual_seed(SEED)
device=torch.device("cuda" if torch.cuda.is_available() else "cpu")

def load(ds):
    tr,va,te=(ds(split=s,download=True) for s in ["train","val","test"])
    def prep(d): return d.imgs.reshape(len(d.imgs),-1).astype(np.float32)/255.0, d.labels.flatten().astype(np.int64)
    return prep(tr),prep(va),prep(te)
def standardise(Xtr,*rest):
    m,s=Xtr.mean(0),Xtr.std(0)+1e-8
    return [(X-m)/s for X in (Xtr,*rest)]
(Xp_tr,yp_tr),(Xp_va,yp_va),(Xp_te,yp_te)=load(PneumoniaMNIST)
(Xb_tr,yb_tr),(Xb_va,yb_va),(Xb_te,yb_te)=load(BloodMNIST)
Xp_tr,Xp_va,Xp_te=standardise(Xp_tr,Xp_va,Xp_te)
Xb_tr,Xb_va,Xb_te=standardise(Xb_tr,Xb_va,Xb_te)

class MLP(nn.Module):
    def __init__(self,in_dim,n,h1=512,h2=256,p=0.3):
        super().__init__(); self.fc1=nn.Linear(in_dim,h1); self.fc2=nn.Linear(h1,h2); self.fc3=nn.Linear(h2,n); self.drop=nn.Dropout(p)
    def forward(self,x):
        x=self.drop(torch.relu(self.fc1(x))); x=self.drop(torch.relu(self.fc2(x))); return self.fc3(x)

def train_mlp(Xtr,ytr,Xva,yva,in_dim,n,batch_size=128,lr=1e-3,dropout=0.3,max_epochs=60,patience=10):
    torch.manual_seed(SEED)
    m=MLP(in_dim,n,p=dropout).to(device); opt=optim.Adam(m.parameters(),lr=lr,weight_decay=1e-4); lf=nn.CrossEntropyLoss()
    ld=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=batch_size,shuffle=True)
    Xv=torch.tensor(Xva).to(device); best,bs,w=0,None,0
    for e in range(max_epochs):
        m.train()
        for xb,yb in ld:
            xb,yb=xb.to(device),yb.to(device); opt.zero_grad(); lf(m(xb),yb).backward(); opt.step()
        m.eval()
        with torch.no_grad(): va=accuracy_score(yva,m(Xv).argmax(1).cpu().numpy())
        if va>best: best,bs,w=va,copy.deepcopy(m.state_dict()),0
        else:
            w+=1
            if w>=patience: break
    m.load_state_dict(bs); return m

def evaluate(m,X,y):
    m.eval()
    with torch.no_grad(): p=m(torch.tensor(X).to(device)).argmax(1).cpu().numpy()
    return accuracy_score(y,p), f1_score(y,p,average='macro')

def prune_model(m,a):
    pm=copy.deepcopy(m)
    for _,mod in pm.named_modules():
        if isinstance(mod,nn.Linear): prune.l1_unstructured(mod,'weight',amount=a); prune.remove(mod,'weight')
    return pm
def quantize(m,bits):
    qm=copy.deepcopy(m)
    with torch.no_grad():
        for mod in qm.modules():
            if isinstance(mod,nn.Linear):
                w=mod.weight.data; lo,hi=w.min(),w.max()
                if hi>lo: lv=2**bits-1; sc=(hi-lo)/lv; mod.weight.copy_(torch.round((w-lo)/sc)*sc+lo)
    return qm
def finetune(m,Xtr,ytr,epochs=10,lr=5e-4,bs=128):
    m.train(); opt=optim.Adam(m.parameters(),lr=lr,weight_decay=1e-4); lf=nn.CrossEntropyLoss()
    ld=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=bs,shuffle=True)
    for _ in range(epochs):
        for xb,yb in ld:
            xb,yb=xb.to(device),yb.to(device); opt.zero_grad(); lf(m(xb),yb).backward(); opt.step()
    return m
def finetune_keep_quant(m,Xtr,ytr,bits,epochs=10,lr=5e-4,bs=128):
    m.train(); opt=optim.Adam(m.parameters(),lr=lr,weight_decay=1e-4); lf=nn.CrossEntropyLoss()
    ld=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=bs,shuffle=True)
    for _ in range(epochs):
        for xb,yb in ld:
            xb,yb=xb.to(device),yb.to(device); opt.zero_grad(); lf(m(xb),yb).backward(); opt.step()
        with torch.no_grad():
            for mod in m.modules():
                if isinstance(mod,nn.Linear):
                    w=mod.weight.data; lo,hi=w.min(),w.max()
                    if hi>lo: lv=2**bits-1; sc=(hi-lo)/lv; mod.weight.copy_(torch.round((w-lo)/sc)*sc+lo)
    return m

def fair_PtoQ(m,Xtr,ytr,a,bits):
    pm=prune_model(m,a); pm=finetune(pm,Xtr,ytr); pm=quantize(pm,bits); return finetune_keep_quant(pm,Xtr,ytr,bits)
def fair_QtoP(m,Xtr,ytr,a,bits):
    qm=quantize(m,bits); qm=finetune_keep_quant(qm,Xtr,ytr,bits); pm=prune_model(qm,a); return finetune_keep_quant(pm,Xtr,ytr,bits)
def fair_order_study(m,Xtr,ytr,Xte,yte,name,combos):
    print(f"\n{name} — FAIR order (P->Q vs Q->P)")
    print(f"{'prune':>7}{'bits':>6}{'P->Q':>9}{'Q->P':>9}{'diff':>8}")
    rows=[]
    for a,bits in combos:
        x1=evaluate(fair_PtoQ(m,Xtr,ytr,a,bits),Xte,yte)[0]; x2=evaluate(fair_QtoP(m,Xtr,ytr,a,bits),Xte,yte)[0]
        rows.append((a,bits,x1,x2,x1-x2)); print(f"{a:>6.0%}{bits:>6}{x1:>9.3f}{x2:>9.3f}{x1-x2:>+8.3f}")
    return rows

mlp_p=train_mlp(Xp_tr,yp_tr,Xp_va,yp_va,784,2,dropout=0.5,lr=3e-4)
mlp_b=train_mlp(Xb_tr,yb_tr,Xb_va,yb_va,2352,8,dropout=0.3,lr=1e-3)
print("baselines ready:", evaluate(mlp_p,Xp_te,yp_te)[0].round(3), evaluate(mlp_b,Xb_te,yb_te)[0].round(3))
```

```python
def batch_size_study(Xtr,ytr,Xva,yva,Xte,yte,in_dim,n,name,dropout,lr,
                     prune_amt=0.5,bits=4,batch_sizes=(32,64,128,256)):
    print(f"\n{name} — batch size: uncompressed vs compressed")
    print(f"{'batch':>7}{'uncomp':>9}{'comp':>9}")
    rows=[]
    for bs in batch_sizes:
        # uncompressed: train fresh at this batch size
        m=train_mlp(Xtr,ytr,Xva,yva,in_dim,n,batch_size=bs,dropout=dropout,lr=lr)
        unc=evaluate(m,Xte,yte)[0]
        # compressed: prune + finetune (at this batch size) + quantize
        pm=prune_model(m,prune_amt)
        pm=finetune(pm,Xtr,ytr,bs=bs)
        pm=quantize(pm,bits)
        comp=evaluate(pm,Xte,yte)[0]
        rows.append((bs,unc,comp)); print(f"{bs:>7}{unc:>9.3f}{comp:>9.3f}")
    return rows

bs_p=batch_size_study(Xp_tr,yp_tr,Xp_va,yp_va,Xp_te,yp_te,784,2,"Pneumonia",0.5,3e-4)
bs_b=batch_size_study(Xb_tr,yb_tr,Xb_va,yb_va,Xb_te,yb_te,2352,8,"Blood",0.3,1e-3)
```

```python
# ============================================================
# Path A: QAT in the COMBINED setting (prune + low-bit quant)
# Compares prune->PTQ (broke before) vs prune->QAT (the fix)
# ============================================================
def prune_then_ptq(model, Xtr, ytr, amount, bits):
    # prune, fine-tune, then post-training quantize (the version that broke)
    pm = copy.deepcopy(model)
    for _,m in pm.named_modules():
        if isinstance(m,nn.Linear): prune.l1_unstructured(m,'weight',amount=amount); prune.remove(m,'weight')
    # quick fine-tune
    pm.train(); opt=optim.Adam(pm.parameters(),lr=5e-4,weight_decay=1e-4); lf=nn.CrossEntropyLoss()
    ld=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=128,shuffle=True)
    for _ in range(10):
        for xb,yb in ld:
            xb,yb=xb.to(device),yb.to(device); opt.zero_grad(); lf(pm(xb),yb).backward(); opt.step()
    return quantize(pm, bits)

def prune_then_qat(model, Xtr, ytr, Xva, yva, in_dim, n, amount, bits, dropout, lr):
    # prune a QAT model, then train it quantization-aware (the fix)
    torch.manual_seed(SEED)
    m = QATMLP(in_dim, n, bits, p=dropout).to(device)
    # copy pruned baseline weights in as a starting point
    m.fc1.weight.data = model.fc1.weight.data.clone()
    m.fc2.weight.data = model.fc2.weight.data.clone()
    m.fc3.weight.data = model.fc3.weight.data.clone()
    # prune the QAT model's layers
    for _,mod in m.named_modules():
        if isinstance(mod,nn.Linear): prune.l1_unstructured(mod,'weight',amount=amount); prune.remove(mod,'weight')
    # train quantization-aware
    opt=optim.Adam(m.parameters(),lr=lr,weight_decay=1e-4); lf=nn.CrossEntropyLoss()
    ld=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=128,shuffle=True)
    Xv=torch.tensor(Xva).to(device); best,bs_,w=0,None,0
    for e in range(40):
        m.train()
        for xb,yb in ld:
            xb,yb=xb.to(device),yb.to(device); opt.zero_grad(); lf(m(xb),yb).backward(); opt.step()
        m.eval()
        with torch.no_grad(): va=accuracy_score(yva,m(Xv).argmax(1).cpu().numpy())
        if va>best: best,bs_,w=va,copy.deepcopy(m.state_dict()),0
        else:
            w+=1
            if w>=10: break
    m.load_state_dict(bs_); return m

print("Path A: prune + low-bit quant — PTQ (old) vs QAT (improvement)")
print(f"{'dataset':>10}{'prune':>7}{'bits':>6}{'PTQ':>9}{'QAT':>9}{'gain':>8}")
configs = {
    "Pneumonia":(Xp_tr,yp_tr,Xp_va,yp_va,Xp_te,yp_te,784,2,0.5,3e-4,mlp_p),
    "Blood":(Xb_tr,yb_tr,Xb_va,yb_va,Xb_te,yb_te,2352,8,0.3,1e-3,mlp_b),
}
for name,(Xtr,ytr,Xva,yva,Xte,yte,ind,nc,dr,lr,base) in configs.items():
    for amount,bits in [(0.7,3),(0.9,3),(0.7,2),(0.9,2)]:
        ptq=evaluate(prune_then_ptq(base,Xtr,ytr,amount,bits),Xte,yte)[0]
        qm=prune_then_qat(base,Xtr,ytr,Xva,yva,ind,nc,amount,bits,dr,lr)
        qat=evaluate(qm,Xte,yte)[0]
        print(f"{name:>10}{amount:>6.0%}{bits:>6}{ptq:>9.3f}{qat:>9.3f}{qat-ptq:>+8.3f}")
```

```python
!pip install medmnist -q
import numpy as np, torch, copy, warnings
import torch.nn as nn, torch.optim as optim
import torch.nn.functional as F
import torch.nn.utils.prune as prune
from torch.utils.data import DataLoader, TensorDataset
from medmnist import PneumoniaMNIST, BloodMNIST
from sklearn.metrics import accuracy_score, f1_score
warnings.filterwarnings("ignore")

SEED=42; np.random.seed(SEED); torch.manual_seed(SEED)
device=torch.device("cuda" if torch.cuda.is_available() else "cpu")

def load(ds):
    tr,va,te=(ds(split=s,download=True) for s in ["train","val","test"])
    def prep(d): return d.imgs.reshape(len(d.imgs),-1).astype(np.float32)/255.0, d.labels.flatten().astype(np.int64)
    return prep(tr),prep(va),prep(te)
def standardise(Xtr,*rest):
    m,s=Xtr.mean(0),Xtr.std(0)+1e-8
    return [(X-m)/s for X in (Xtr,*rest)]
(Xp_tr,yp_tr),(Xp_va,yp_va),(Xp_te,yp_te)=load(PneumoniaMNIST)
(Xb_tr,yb_tr),(Xb_va,yb_va),(Xb_te,yb_te)=load(BloodMNIST)
Xp_tr,Xp_va,Xp_te=standardise(Xp_tr,Xp_va,Xp_te)
Xb_tr,Xb_va,Xb_te=standardise(Xb_tr,Xb_va,Xb_te)

class MLP(nn.Module):
    def __init__(self,in_dim,n,h1=512,h2=256,p=0.3):
        super().__init__(); self.fc1=nn.Linear(in_dim,h1); self.fc2=nn.Linear(h1,h2); self.fc3=nn.Linear(h2,n); self.drop=nn.Dropout(p)
    def forward(self,x):
        x=self.drop(torch.relu(self.fc1(x))); x=self.drop(torch.relu(self.fc2(x))); return self.fc3(x)

def evaluate(m,X,y):
    m.eval()
    with torch.no_grad(): p=m(torch.tensor(X).to(device)).argmax(1).cpu().numpy()
    return accuracy_score(y,p), f1_score(y,p,average='macro')
def quantize(m,bits):
    qm=copy.deepcopy(m)
    with torch.no_grad():
        for mod in qm.modules():
            if isinstance(mod,nn.Linear):
                w=mod.weight.data; lo,hi=w.min(),w.max()
                if hi>lo: lv=2**bits-1; sc=(hi-lo)/lv; mod.weight.copy_(torch.round((w-lo)/sc)*sc+lo)
    return qm

class FakeQuant(torch.autograd.Function):
    @staticmethod
    def forward(ctx,w,bits):
        lo,hi=w.min(),w.max()
        if hi==lo: return w
        lv=2**bits-1; sc=(hi-lo)/lv
        return torch.round((w-lo)/sc)*sc+lo
    @staticmethod
    def backward(ctx,g): return g,None
def fake_quant(w,bits): return FakeQuant.apply(w,bits)

class QATMLP(nn.Module):
    def __init__(self,in_dim,n,bits,h1=512,h2=256,p=0.3):
        super().__init__(); self.fc1=nn.Linear(in_dim,h1); self.fc2=nn.Linear(h1,h2); self.fc3=nn.Linear(h2,n); self.drop=nn.Dropout(p); self.bits=bits
    def forward(self,x):
        w1=fake_quant(self.fc1.weight,self.bits); x=self.drop(torch.relu(F.linear(x,w1,self.fc1.bias)))
        w2=fake_quant(self.fc2.weight,self.bits); x=self.drop(torch.relu(F.linear(x,w2,self.fc2.bias)))
        w3=fake_quant(self.fc3.weight,self.bits); return F.linear(x,w3,self.fc3.bias)

from google.colab import drive
drive.mount('/content/drive')
import os
SAVE_DIR='/content/drive/MyDrive/ml2026_project'
mlp_p=MLP(784,2,p=0.5).to(device); mlp_p.load_state_dict(torch.load(f"{SAVE_DIR}/mlp_p.pt")); mlp_p.eval()
mlp_b=MLP(2352,8,p=0.3).to(device); mlp_b.load_state_dict(torch.load(f"{SAVE_DIR}/mlp_b.pt")); mlp_b.eval()
print("ready:", round(evaluate(mlp_p,Xp_te,yp_te)[0],3), round(evaluate(mlp_b,Xb_te,yb_te)[0],3))
```

```python
# ============================================================
# QAT improvement with ERROR BARS (3 seeds) — headline result
# Run after the Path A setup block (needs QATMLP, quantize, mlp_p, mlp_b)
# ============================================================
import numpy as np

def prune_then_ptq_seeded(model, Xtr, ytr, amount, bits, seed):
    torch.manual_seed(seed); np.random.seed(seed)
    pm = copy.deepcopy(model)
    for _,m in pm.named_modules():
        if isinstance(m,nn.Linear): prune.l1_unstructured(m,'weight',amount=amount); prune.remove(m,'weight')
    pm.train(); opt=optim.Adam(pm.parameters(),lr=5e-4,weight_decay=1e-4); lf=nn.CrossEntropyLoss()
    ld=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=128,shuffle=True)
    for _ in range(10):
        for xb,yb in ld:
            xb,yb=xb.to(device),yb.to(device); opt.zero_grad(); lf(pm(xb),yb).backward(); opt.step()
    return quantize(pm, bits)

def prune_then_qat_seeded(model, Xtr, ytr, Xva, yva, in_dim, n, amount, bits, dropout, lr, seed):
    torch.manual_seed(seed); np.random.seed(seed)
    m = QATMLP(in_dim, n, bits, p=dropout).to(device)
    m.fc1.weight.data = model.fc1.weight.data.clone()
    m.fc2.weight.data = model.fc2.weight.data.clone()
    m.fc3.weight.data = model.fc3.weight.data.clone()
    for _,mod in m.named_modules():
        if isinstance(mod,nn.Linear): prune.l1_unstructured(mod,'weight',amount=amount); prune.remove(mod,'weight')
    opt=optim.Adam(m.parameters(),lr=lr,weight_decay=1e-4); lf=nn.CrossEntropyLoss()
    ld=DataLoader(TensorDataset(torch.tensor(Xtr),torch.tensor(ytr)),batch_size=128,shuffle=True)
    Xv=torch.tensor(Xva).to(device); best,bs_,w=0,None,0
    for e in range(40):
        m.train()
        for xb,yb in ld:
            xb,yb=xb.to(device),yb.to(device); opt.zero_grad(); lf(m(xb),yb).backward(); opt.step()
        m.eval()
        with torch.no_grad(): va=accuracy_score(yva,m(Xv).argmax(1).cpu().numpy())
        if va>best: best,bs_,w=va,copy.deepcopy(m.state_dict()),0
        else:
            w+=1
            if w>=10: break
    m.load_state_dict(bs_); return m

SEEDS=[42,43,44]
print("QAT improvement with error bars (mean ± std over 3 seeds)")
print(f"{'dataset':>10}{'prune':>7}{'bits':>6}{'PTQ':>16}{'QAT':>16}")
qat_eb_rows=[]
configs = {
    "Pneumonia":(Xp_tr,yp_tr,Xp_va,yp_va,Xp_te,yp_te,784,2,0.5,3e-4,mlp_p),
    "Blood":(Xb_tr,yb_tr,Xb_va,yb_va,Xb_te,yb_te,2352,8,0.3,1e-3,mlp_b),
}
for name,(Xtr,ytr,Xva,yva,Xte,yte,ind,nc,dr,lr,base) in configs.items():
    for amount,bits in [(0.7,3),(0.9,3),(0.7,2),(0.9,2)]:
        ptqs=[evaluate(prune_then_ptq_seeded(base,Xtr,ytr,amount,bits,s),Xte,yte)[0] for s in SEEDS]
        qats=[evaluate(prune_then_qat_seeded(base,Xtr,ytr,Xva,yva,ind,nc,amount,bits,dr,lr,s),Xte,yte)[0] for s in SEEDS]
        pm_,ps_=np.mean(ptqs),np.std(ptqs); qm_,qs_=np.mean(qats),np.std(qats)
        qat_eb_rows.append((name,amount,bits,pm_,ps_,qm_,qs_))
        print(f"{name:>10}{amount:>6.0%}{bits:>6}   {pm_:.3f}±{ps_:.3f}   {qm_:.3f}±{qs_:.3f}")

import pandas as pd
pd.DataFrame(qat_eb_rows,columns=["dataset","prune","bits","PTQ_mean","PTQ_std","QAT_mean","QAT_std"]).to_csv(f"{SAVE_DIR}/qat_errorbars.csv",index=False)
print("saved with error bars")
```
